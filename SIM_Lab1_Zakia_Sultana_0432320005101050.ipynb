```python
#Task:01
import numpy as np

scalar = 3.7
arr = np.array([0.0, 0.5, 1.0, 1.5, 2.0, 2.5])

add_result = arr + scalar
sub_result = arr - scalar
mul_result = arr * scalar
div_result = arr / scalar
pow_result = arr ** scalar

sin_vals = np.sin(arr)
cos_vals = np.cos(arr)
tan_vals = np.tan(arr)
exp_vals = np.exp(arr)
sqrt_vals = np.sqrt(arr)
log_vals = np.log(arr + 1)

print("Addition:", add_result)
print("Subtraction:", sub_result)
print("Multiplication:", mul_result)
print("Division:", div_result)
print("Power:", pow_result)

print("Sin:", sin_vals)
print("Cos:", cos_vals)
print("Tan:", tan_vals)
print("Exp:", exp_vals)
print("Sqrt:", sqrt_vals)
print("Log:", log_vals)
```

```python
#Task:03
import numpy as np

# 2x3 matrix with custom float values
matrix = np.array([
    [2.5, 4.7, 1.2],
    [8.3, 3.6, 6.1]
])

# max and min with index
max_val = np.max(matrix)
min_val = np.min(matrix)
max_idx = np.unravel_index(np.argmax(matrix), matrix.shape)
min_idx = np.unravel_index(np.argmin(matrix), matrix.shape)

# vector length
vector_length = np.linalg.norm(matrix)

# sorting (row-wise and full matrix)
sorted_row_desc = -np.sort(-matrix, axis=1)
sorted_entire_desc = np.sort(matrix.flatten())[::-1]

# sum and product
sum_val = np.sum(matrix)
prod_val = np.prod(matrix)

# mean, median and std
median_val = np.median(matrix)
mean_val = np.mean(matrix)
std_val = np.std(matrix)

# print outputs
print("Original Matrix (2x3):\n", matrix)

print("\n--- Max & Min ---")
print("Max Value:", max_val, "at index:", max_idx)
print("Min Value:", min_val, "at index:", min_idx)

print("\n--- Vector Length ---")
print("Vector Length:", vector_length)

print("\n--- Sorting ---")
print("Row-wise Descending:\n", sorted_row_desc)
print("Entire Matrix Descending:", sorted_entire_desc)

print("\n--- Sum & Product ---")
print("Sum:", sum_val)
print("Product:", prod_val)

print("\n--- Statistics ---")
print("Median:", median_val)
print("Mean:", mean_val)
print("Standard Deviation:", std_val)
```

```python
#Task:02
import numpy as np

# Create 3x2 matrix
matrix = np.array([
    [0.0, 0.5],
    [-1.5, 2.7],
    [3.2, -0.8]
])

# Clip values between -1 and 1 for arcsin/arccos
matrix_clipped = np.clip(matrix, -1.0, 1.0)

# Positive values for log and sqrt
matrix_positive = np.abs(matrix) + 1e-5

# Trigonometric functions
sin_res = np.sin(matrix)
cos_res = np.cos(matrix)
tan_res = np.tan(matrix)
arcsin_res = np.arcsin(matrix_clipped)
arccos_res = np.arccos(matrix_clipped)
arctan_res = np.arctan(matrix)

# Exponential and Logarithmic functions
exp_res = np.exp(matrix)
log_res = np.log(matrix_positive)
log10_res = np.log10(matrix_positive)

# Absolute value and Square root
abs_res = np.abs(matrix)
sqrt_res = np.sqrt(matrix_positive)

# Remainder when divided by 2
remainder_res = np.remainder(matrix, 2)

# Rounding operations
round_res = np.round(matrix)
floor_res = np.floor(matrix)
ceil_res = np.ceil(matrix)

# Print results
print("Original Matrix (3x2):\n", matrix)

print("\n--- Trigonometric Functions ---")
print("Sin:\n", sin_res)
print("Cos:\n", cos_res)
print("Tan:\n", tan_res)
print("Arcsin:\n", arcsin_res)
print("Arccos:\n", arccos_res)
print("Arctan:\n", arctan_res)

print("\n--- Exponential & Logarithmic Functions ---")
print("Exp:\n", exp_res)
print("Log:\n", log_res)
print("Log10:\n", log10_res)

print("\n--- Absolute & Square Root ---")
print("Absolute Value:\n", abs_res)
print("Square Root:\n", sqrt_res)

print("\n--- Remainder ---")
print("Remainder (divided by 2):\n", remainder_res)

print("\n--- Rounding ---")
print("Round:\n", round_res)
print("Floor:\n", floor_res)
print("Ceil:\n", ceil_res)
```
