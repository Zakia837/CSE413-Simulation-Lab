```python
import numpy as np
import matplotlib.pyplot as plt



# Part 1 — Data Generation

# 1. Generate 3,000 wait times for regular days (Uniform Distribution: 2 to 12 mins)
regular_days = np.random.uniform(low=2.0, high=12.0, size=3000)

# 2. Generate 3,000 wait times for busy days (Normal Distribution: Mean=8, Std Dev=2)
busy_days = np.random.normal(loc=8.0, scale=2.0, size=3000)

# 3. Ensure no negative wait times (replace negative values with 0)
busy_days = np.clip(busy_days, a_min=0, a_max=None)


# Part 2 — Visualization


# Option: Two subplots in the same figure for clear comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

# Histogram for Regular Days
axes[0].hist(regular_days, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
axes[0].set_title("Regular Days (Uniform Distribution)")
axes[0].set_xlabel("Wait Time (minutes)")
axes[0].set_ylabel("Frequency")
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

# Histogram for Busy Days
axes[1].hist(busy_days, bins=30, color='coral', edgecolor='black', alpha=0.7)
axes[1].set_title("Busy Days (Normal Distribution)")
axes[1].set_xlabel("Wait Time (minutes)")
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.suptitle("Customer Wait Times Comparison", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Part 3 — Statistical Analysis


def calculate_stats(data, label):
    mean_val = np.mean(data)
    median_val = np.median(data)
    std_val = np.std(data)
    max_val = np.max(data)

    print(f"--- {label} ---")
    print(f"Mean:               {mean_val:.2f} mins")
    print(f"Median:             {median_val:.2f} mins")
    print(f"Standard Deviation: {std_val:.2f} mins")
    print(f"Maximum Wait Time:  {max_val:.2f} mins\n")

# Print statistical summary
calculate_stats(regular_days, "Regular Days (Uniform)")
calculate_stats(busy_days, "Busy Days (Normal)")
```
