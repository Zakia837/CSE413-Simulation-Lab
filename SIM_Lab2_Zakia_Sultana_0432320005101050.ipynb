```python

import numpy as np

# Task 1: Personal Matrix Generation
# Student ID: 0432320005101050
d1, d2 = 0, 5

A = np.array([[d1 + 2, d2 + 1],
              [2 * d1, d2 + 2]])

print("Matrix A:\n", A)

#  Task 2: Analyze Matrix A
print("\n Task 2 Results")
print("Shape of A:", A.shape)
print("Determinant:", np.linalg.det(A))
print("Rank:", np.linalg.matrix_rank(A))
print("Eigenvalues:", np.linalg.eigvals(A))
print("Inverse of A:\n", np.linalg.inv(A))
```

```python
import numpy as np

# Task 3: Value Change Experiment
# Changing top-left element
B = np.array([[3, 6],
              [0, 7]])

print("Matrix B:\n", B)

#  Task 4: Re-analyze Matrix B
print("\n Task 4 Results")
print("Shape of B:", B.shape)
print("Determinant:", np.linalg.det(B))
print("Rank:", np.linalg.matrix_rank(B))
print("Eigenvalues:", np.linalg.eigvals(B))
print("Inverse of B:\n", np.linalg.inv(B))
```

```python
1.How did the determinant change and why?
The determinant increased from 14 to 21. Since A and B are upper triangular, \det = a_{11} \times a_{22}. Increasing a_{11} from 2 to 3 increased the determinant by 1 \times 7 = 7.

2.Did the rank change? Explain.
No, the rank stayed 2. Both determinants are non-zero (\det(A)=14, \det(B)=21), meaning both matrices maintain full linear independence.

3.How did the eigenvalues respond to the value change?
Because both matrices are upper triangular, the eigenvalues are simply the diagonal entries. Changing a_{11} from 2 to 3 changed the first eigenvalue from 2 to 3, while the second remained 7.

4.Is B easier or harder to invert than A? Why?
Matrix B is slightly easier (more numerically stable) to invert because its determinant is larger (21 vs 14), pushing it further away from being a singular matrix (\det = 0).
```
